**GOLD LAYER**

DIMDATE TABLE

```python
df_silver_customers=spark.read.format("delta").\
    load("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Silver/silver_customers")
df_silver_stores=spark.read.format("delta").\
    load("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Silver/silver_stores")
df_silver_products=spark.read.format("delta").\
    load("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Silver/silver_products")
df_silver_transactions=spark.read.format("delta").\
    load("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Silver/silver_transactions")
print(df_silver_customers.dtypes)
print(df_silver_stores.dtypes)
print(df_silver_products.dtypes)
print(df_silver_transactions.dtypes)
```

```python
from pyspark.sql.functions import col, date_format,dayofmonth, dayofweek, month, quarter, year

df_cust_dates = df_silver_customers.select("registration_date")
df_trn_dates = df_silver_transactions.select("transaction_date")
df_dates = df_cust_dates.union(df_trn_dates).distinct().sort(col("registration_date")).withColumnRenamed("registration_date", "Date")
df_dates = df_dates.withColumn("DateKey",date_format(col("Date"), "yyyyMMdd").cast("int"))
display(df_dates)
```

```python
df_gold_dimDate = df_dates.select("DateKey","Date").\
    withColumn("Day", dayofmonth(col("Date"))).\
    withColumn("Month", month(col("Date"))).\
    withColumn("Quarter", quarter(col("Date"))).\
    withColumn("Year", year(col("Date"))).\
    withColumn("Weekday", dayofweek(col("Date")))
display(df_gold_dimDate)
```

```python
df_gold_dimDate.write.format("delta").mode("overwrite").\
    save("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Gold/DimDate")

df_gold_dimDate.write.format("delta").mode("overwrite").saveAsTable("DimDate")
```

DIMCUSTOMERS

```python
from pyspark.sql.functions import monotonically_increasing_id

df_gold_dimCustomers= df_silver_customers.alias("c")\
    .join(df_gold_dimDate.alias("d"), col("c.registration_date") == col("d.Date"))\
    .select("c.customer_id", "c.first_name", "c.last_name", "c.email", "c.phone", "c.city","d.DateKey").\
    withColumn("CustomerKey", monotonically_increasing_id()+1).\
    withColumnRenamed("customer_id", "CustomerID").\
    withColumnRenamed("first_name", "FirstName").\
    withColumnRenamed("last_name", "LastName").\
    withColumnRenamed("email", "Email").\
    withColumnRenamed("phone", "Phone").\
    withColumnRenamed("city", "City")
display(df_gold_dimCustomers)
```

```python
df_gold_dimCustomers=df_gold_dimCustomers\
    .select("CustomerKey", "CustomerID", "FirstName", "LastName", "Email", "Phone", "City","DateKey")
```

```python
df_gold_dimCustomers.write.format("delta").mode("overwrite").\
    save("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Gold/DimCustomers")

df_gold_dimCustomers.write.format("delta").mode("overwrite").saveAsTable("DimCustomers")
```

DIMSTORES

```python
df_gold_dimStores= df_silver_stores.select("store_id", "store_name", "location").\
    withColumn("StoreKey", monotonically_increasing_id()+1).\
    withColumnRenamed("store_id", "StoreID").\
    withColumnRenamed("store_name", "StoreName").\
    withColumnRenamed("location", "Location")
                                           
display(df_gold_dimStores)
```

```python
df_gold_dimStores=df_gold_dimStores.select("StoreKey", "StoreID", "StoreName","Location")
```

```python
df_gold_dimStores.write.format("delta").mode("overwrite").\
    save("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Gold/DimStores")

df_gold_dimStores.write.format("delta").mode("overwrite").saveAsTable("DimStores")
```

DIMPRODUCTS

```python
df_gold_dimProducts= df_silver_products.select("product_id", "product_name", "category", "price").\
    withColumn("ProductKey", monotonically_increasing_id()+1).\
    withColumnRenamed("product_id", "ProductID").\
    withColumnRenamed("product_name", "ProductName").\
    withColumnRenamed("category", "Category").\
    withColumnRenamed("price", "Price")
display(df_gold_dimProducts)
```

```python
df_gold_dimProducts=df_gold_dimProducts.select("ProductKey", "ProductID", "ProductName", "Category", "Price")
```

```python
df_gold_dimProducts.write.format("delta").mode("overwrite").\
    save("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Gold/DimProducts")

df_gold_dimProducts.write.format("delta").mode("overwrite").saveAsTable("DimProducts")
```

FACTTRANSACTIONS

```python
df_gold_factTransactions= df_silver_transactions.alias("t")\
    .join(df_gold_dimCustomers.alias("c"), col("t.customer_id") == col("c.CustomerID"))\
    .join(df_gold_dimProducts.alias("p"), col("t.product_id") == col("p.ProductID"))\
    .join(df_gold_dimStores .alias("s"), col("t.store_id") == col("s.StoreID"))\
    .join(df_gold_dimDate.alias("d"), col("t.transaction_date") == col("d.Date"))\
    .select(
        col("t.transaction_id"),
        col("c.CustomerKey"),
        col("p.ProductKey"),
        col("s.StoreKey"),
        col("d.DateKey"),
        col("t.quantity"),
        col("p.Price"))\
    .withColumn("SalesAmount", col("p.Price") * col("t.quantity"))\
    .withColumnRenamed("transaction_id", "TransactionID")\
    .withColumnRenamed("quantity", "Quantity")\
    .drop("Price")
    

display(df_gold_factTransactions)
```

```python

df_gold_factTransactions.write.format("delta").mode("overwrite").\
    save("abfss://destination@projectstoragesruthy.dfs.core.windows.net/Gold/FactTransactions")
    
df_gold_factTransactions.write.format("delta").mode("overwrite").saveAsTable("FactTransactions")
```
